```python
import numpy as np
import matplotlib.pyplot as plt #import the appropriate modules for mathematical and plotting purpose

def f(x):
    return x**5+12.5 #define the function
def df(x):
    return 5*x**4    #its derivative

x=1                #define the value of x to plot the result
true_val=df(x)    #we define the true value as the derivative

h=np.logspace(-6,-1,100)  #defining the range of our iterations (10^-6 to 10^-1) into 100 divisions

diff_f=(f(x+h)-f(x))/h    #forward difference
diff_c=(f(x+h/2)-f(x-h/2))/h  #central difference
diff_complex=np.imag(f(x+1j*h))/h  #complex step method

#plotting functions
plt.plot(h,diff_f,color="red",label="forward difference") #plot the forward function
plt.plot(h,diff_c,color="blue",label="central difference") #plot the central function
plt.plot(h,diff_complex,color="green",label="complex step") #plot the complex step function
plt.axhline(true_val,color="black",linestyle="dotted",label="true derivative")  #an x-axis line showing actual value of derivative
plt.xlabel("Step size h") 
plt.xscale('log') #value of x-axis in terms of log
plt.ylabel("Derivative value")
plt.title("Derivative value vs step size h")
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return (x-4)**2-1 #defining the function

def bisection(a,b,tol,max_iter,real_root=5):  #defining our bisection function, we got [a,b] range, tolerance on how close the value can be, maximum number of iterations and the actual root
    Xr=[]      #aprroximated root array
    iters=[]   #iterations array
    diff=[]    #difference between new and old approx. root array
    error=[]   #absolute error between approx root and actual
    
    xr_old=(a+b)/2  #defining old approximated root
    
    #for loop to keep the operations going till either the desired tolerance reached or maximum iteration exhuasted
    for i in range(max_iter):  #gives out an input of number of iterations which is 100 here, we get 0-100 iterations if tolerance not satisfied
        xr=(a+b)/2  #new approx root
        print(f"The root is {xr} at {i+1} iterations")
        
        Xr.append(xr)   #append function, it stores new data and stores it in the array defined one after another as loop goes on
        iters.append(i+1)
        diff.append(abs(xr - xr_old))
        error.append(abs(xr-5))
        
        if f(a)*f(xr)<0:      #if function used to substitute a and b with the new approx. roots (xr)
            b=xr     
        elif f(a)*f(xr)>0:
            a=xr
        else:
            break          #if f(a)*f(xr)=0 we have reached root
        if abs(a-b) <= tol:   #but approximated answer shoulb enough
            break
        xr_old=xr       

    return xr,Xr,iters,diff,error     #return the values through bisection function

root,Xr,iters,diff,error=bisection(4,10,0.001,100)    #define your values a, b, tolerance and max iteration and the corresponding values we get root, approx. root, difference, error

print(f"Actual root: 5")

#plotting three different plots
plt.plot(iters,Xr,color="green",marker="o")
plt.title("Root vs Iteration")
plt.xlabel("Iterations")
plt.ylabel("Root Value (Xr)")
plt.grid(True)
plt.savefig(r"C:\Users\Apoorva Singh\OneDrive - Amity University\Desktop\ANM Python assignment\Q3-1a.png",dpi=600)   #directly saves the png of graph to file path
plt.show()

plt.plot(iters,diff,color="red",marker="o")
plt.title("|Xi+1-Xi| vs Iteration")
plt.xlabel("Iterations")
plt.ylabel("|Xi+1-Xi|")
plt.grid(True)
plt.savefig(r"C:\Users\Apoorva Singh\OneDrive - Amity University\Desktop\ANM Python assignment\Q3-2a.png",dpi=600)
plt.show()

plt.plot(iters,error,color="black",marker='o')
plt.xlabel("iteration")
plt.ylabel("|Xr-X|")
plt.title("absolute error vs iteration")
plt.grid(True)
plt.savefig(r"C:\Users\Apoorva Singh\OneDrive - Amity University\Desktop\ANM Python assignment\Q3-3a.png",dpi=600)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

y=lambda x: (x-4)**2-1  #the original function
dy=lambda x: 2*(x-4)      #the derivative function

def nraphson(fn,dfn,x,tol,max_iter):  #defining Newton-Raphson function function, derivative, value of x, tolerance and max iteration
    Xr=[]      #array definition
    iteration=[]
    diff=[]
    error=[]
    for i in range (max_iter):
        
        xnew=x-fn(x)/dfn(x)
        
        Xr.append(abs(xnew))
        iteration.append(i+1)
        diff.append(abs(xnew-x))  #relative error
        error.append(abs(xnew-5)) #absolute error
        
        if abs(xnew-x)<tol:
            return xnew, Xr, iteration, diff, error
        x=xnew
        print(f"The root is {xnew} at {i+1} iterations.")

xnew,Xr,iteration,diff,error=nraphson(y,dy,6,0.001,100) 

print(f"with actual root as 5")

plt.plot(iteration,Xr,color="green",marker="o")
plt.title("Root vs Iteration")
plt.xlabel("Iterations")
plt.ylabel("Root Value (Xr)")
plt.grid(True)
plt.savefig(r"C:\Users\Apoorva Singh\OneDrive - Amity University\Desktop\ANM Python assignment\Q4-1a.png",dpi=600)
plt.show()

plt.plot(iteration,diff,color="red",marker="o")
plt.title("|Xi+1-Xi| vs Iteration")
plt.xlabel("Iterations")
plt.ylabel("|Xi+1-Xi|")
plt.grid(True)
plt.savefig(r"C:\Users\Apoorva Singh\OneDrive - Amity University\Desktop\ANM Python assignment\Q4-1b.png",dpi=600)
plt.show()

plt.plot(iteration,error,color="black",marker="o")
plt.title("absolute error vs Iteration")
plt.xlabel("Iterations")
plt.ylabel("|Xr-Xi|")
plt.grid(True)
plt.savefig(r"C:\Users\Apoorva Singh\OneDrive - Amity University\Desktop\ANM Python assignment\Q4-1c.png",dpi=600)
plt.show()
```
