```python
#Surface plot not contour, got it from youtube neural nine
import numpy as np
import matplotlib.pyplot as plt

def z_function(x,y):
    return (x-2)**2+(y-4)**2

def grad(x,y):
    return 2*(x-2),2*(y-4)

x=np.linspace(-4,4,100)
y=np.linspace(-4,4,100)

X,Y=np.meshgrid(x,y)
Z=z_function(X,Y)

current_pos=[-10.0,-10.0]

learning_rate=0.001

ax=plt.subplot(projection="3d",computed_zorder=False)

for i in range (1000):
    X_derivative,Y_derivative=grad(current_pos[0],current_pos[1])

    X_new=current_pos[0]-learning_rate*X_derivative
    Y_new=current_pos[1]-learning_rate*Y_derivative

    current_pos = [X_new, Y_new]

ax.plot_surface(X, Y, Z, cmap="plasma", alpha=0.7)

ax.scatter(current_pos[0], current_pos[1], z_function(current_pos[0], current_pos[1]), color="red")
plt.show()
```

```python
# Z-value vs iterations and contour
import numpy as np
import matplotlib.pyplot as plt

def z_function(x,y):
    return (x-2)**2+(y-4)**2 #our function

def grad(x,y):
    return 2*(x-2),2*(y-4)  #gradients of the function

def descent(x,y,a):  #the entire gradient descent process solving
    X=[]
    Y=[]
    iterations=[]
    z_value=[] 
    
    for i in range(100000): #no. of iterations allowed, usually ends before 
        g_x,g_y=grad(x,y)

        x_new=x-a*g_x #our gradient descent formula
        y_new=y-a*g_y

        X.append(x_new)
        Y.append(y_new)
        z_value.append(z_function(x_new,y_new)) #records the z value being gained
        iterations.append(i+1) #records the iterations that takes to solve

        if abs(g_x)<=1e-2 and abs(g_y)<=1e-2: # to verify our result which should be g(x)=0 but since python goes till 16th digit it wont stop.
            return x_new,y_new,i+1,z_value,iterations  

        x,y=x_new,y_new #the next value becomes the previous value as the loop goes on
        
x1,y1,iters1,z_value1,iterations1=descent(0,0,0.01)
print(f"Final solution is {x1,y1} for learning rate=0.01 at {iters1} iterations")  #calling back our descent function for different rates and values

x2,y2,iters2,z_value2,iterations2=descent(0,0,0.001)
print(f"Final solution is {x2,y2} for learning rate=0.001 at {iters2} iterations")

x3,y3,iters3,z_value3,iterations3=descent(0,0,0.004)
print(f"Final solution is {x3,y3} for learning rate=0.004 at {iters3} iterations")

#plotting the z_value vs iterations
plt.plot(iterations1,z_value1,color="blue",label="a=0.01",marker="o",markevery=100) #for learning rate=0.01
plt.plot(iterations2,z_value2,color="red",label="a=0.001",marker="o",markevery=100) #for learning rate=0.001
plt.plot(iterations3,z_value3,color="black",label="a=0.004",marker="o",markevery=100) #for learning rate=0.04
plt.title("z-value vs Iterations")
plt.xlabel("Iterations")
plt.ylabel("z-value")
plt.legend()
plt.grid(True)
plt.savefig(r"C:\Users\Apoorva Singh\OneDrive - Amity University\Desktop\ANM Python assignment\Ass 2\Q1-a1.png",dpi=600)
plt.show()

#plotting the contour simple cmap
x=np.linspace(-20,20,100)
y=np.linspace(-20,20,100)

X,Y=np.meshgrid(x,y)
Z=z_function(X,Y)
plt.contour(X,Y,Z)
plt.colorbar() 
plt.plot(x1,y1,color="blue")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Contour z_function")
plt.savefig(r"C:\Users\Apoorva Singh\OneDrive - Amity University\Desktop\ANM Python assignment\Ass 2\Q1-a2.png",dpi=600)
plt.show()
```

```python
# Z-value vs iterations and contour
import numpy as np
import matplotlib.pyplot as plt

def z_function(x,y):
    return 5*x**2-30*x+y**2-8*y+4*x*y+34

def grad(x,y):
    return 10*x-30+4*y,2*y-8+4*x

def descent(x,y,a):
    X=[]
    Y=[]
    iterations=[]
    z_value=[] 
    
    for i in range(100000):
        g_x,g_y=grad(x,y)

        x_new=x-a*g_x
        y_new=y-a*g_y

        X.append(x_new)
        Y.append(y_new)
        z_value.append(z_function(x_new,y_new))
        iterations.append(i+1)

        if abs(g_x)<1e-2 and abs(g_y)<1e-2:
            return x_new,y_new,i+1,z_value,iterations

        x,y=x_new,y_new

    return x,y,i+1,z_value,iterations

x1,y1,iters1,z_value1,iterations1=descent(0,0,0.05)
print(f"Final solution is {x1,y1} for learning rate=0.05 at {iters1} iterations")

x2,y2,iters2,z_value2,iterations2=descent(0,0,0.02)
print(f"Final solution is {x2,y2} for learning rate=0.02 at {iters2} iterations")

x3,y3,iters3,z_value3,iterations3=descent(0,0,0.1)
print(f"Final solution is {x3,y3} for learning rate=0.1 at {iters3} iterations")

plt.plot(iterations1,z_value1,color="blue",label="a=0.05",marker="o",markevery=50)
plt.plot(iterations2,z_value2,color="red",label="a=0.02",marker="o",markevery=50)
plt.plot(iterations3,z_value3,color="black",label="a=0.1",marker="o",markevery=50)
plt.title("z-value vs Iterations")
plt.xlabel("Iterations")
plt.ylabel("z-value")
plt.grid(True)
plt.legend()
plt.savefig(r"C:\Users\Apoorva Singh\OneDrive - Amity University\Desktop\ANM Python assignment\Ass 2\Q1-b1.png",dpi=600)
plt.show()

x=np.linspace(-30,30,100)
y=np.linspace(-30,30,100)

X,Y=np.meshgrid(x,y)
Z=z_function(X,Y)
plt.contour(X,Y,Z)
plt.colorbar() 
plt.xlabel("x")
plt.ylabel("y")
plt.title("contour z_function")
plt.savefig(r"C:\Users\Apoorva Singh\OneDrive - Amity University\Desktop\ANM Python assignment\Ass 2\Q1-b2.png",dpi=600)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x,y):
    return y**3 #objective function

def g1(x,y):
    return x**2+4*y**2-24  #Constraint function 1
    
def g2(x,y):
    return x+y              #Constraint function 2  

def kkt():    #Starting KKT solving
    solutions=[]
    
    # case l1=l2=0 which means g1 and g2 < 0
    x,y=-np.sqrt(24),0
    if g1(x,y)<=0 and g2(x,y)<=0:
        solutions.append((x,y,0,0,f(x,y)))
        
    #case 2 l1>0 and l2=0
    x,y=0,-np.sqrt(6)
    l1=-(3*y**2/8*y)
    l2=0
    if g1(x,y)<=0 and g2(x,y)<=0 and l1>=0:
        solutions.append((x,y,l1,0,f(x,y)))

    #case 3 l1=0 and l2>0
    x,y=2,-2
    l1 = 0
    l2=-3*y**2
    if g1(x,y)<=0 and g2(x,y)<=0 and l2>=0:
        solutions.append((x,y,0,l2,f(x,y)))

    #case 4 l1>0 and l2>0
    x,y=-y,-np.sqrt(24/5)
    l1=-3*y**2 / (8*y)
    l2=0
    if g1(x,y)<=0 and g2(x,y)<=0 and l2>=0 and l1>=0:
        solutions.append((x,y,l1,l2,f(x,y)))
        
    minimized=min(solutions,key=lambda c:c[4]) #takes the minimum value from the "solutions" list
    return minimized
    
x_value,y_value,l1,l2,value=kkt() #returns x,y value and lambda 1 and 2 along with minimized solutions

print(f"the value of (x,y)= {x_value} and {y_value}")
print(f"value of langrange1 and langrange2= {l1} and {l2}")
print(f"minimized value={value}")

#plotting the functions
X=np.linspace(-6,6,400)
Y=np.linspace(-6,6,400)
X,Y=np.meshgrid(X,Y)  # creates 2D grid for contour
G1=X**2+4*Y**2-24
G2=X+Y

plt.figure(figsize=(6,6))
c1=plt.contour(X,Y,G1,colors="blue") # contour for ellipse
c2=plt.contour(X,Y,G2,colors="green") # contour for line

x_value,y_value,l1,l2,value=kkt()
plt.plot(x_value,y_value,color="red",marker="X",label='min value') #optimal point

plt.title("Constraints vs optimal point")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.savefig(r"C:\Users\Apoorva Singh\OneDrive - Amity University\Desktop\ANM Python assignment\Ass 2\Q3a.png", dpi=600)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return y**3  # objective function

def g1(x, y):
    return x**2+4*y**2-24  # constraint 1

def g2(x, y):
    return x-y  # new constraint 2

def kkt():
    solutions = []
    # case l1=l2=0 -> both constraints inactive
    x,y=-np.sqrt(24),0
    if g1(x,y)<=0 and g2(x,y)<=0:
        solutions.append((x,y,0,0,f(x,y)))
        
    # case l1>0 and l2=0
    x,y=0,-np.sqrt(6)
    l1=-(3*y**2/(8*y)) 
    l2=0
    if g1(x,y)<=0 and g2(x,y)<=0 and l1>=0:
        solutions.append((x,y,l1,l2,f(x, y)))

    x,y=-1,1  
    l1=0
    l2=3*y**2 
    if g1(x,y)<=0 and g2(x,y)<=0 and l2>=0:
        solutions.append((x,y,l1,l2,f(x,y)))

    x,y=-np.sqrt(24/5),-np.sqrt(24/5) 
    l1=-3*y**2 / (8*y)
    l2=0
    if g1(x,y)<=0 and g2(x,y)<=0 and l1>=0 and l2>=0:
        solutions.append((x,y,l1,l2,f(x,y)))
    
    minimized = min(solutions, key=lambda c: c[4])
    return minimized
    
x_value,y_value,l1,l2,value=kkt() #returns x,y value and lambda 1 and 2 along with minimized solutions

print(f"the value of (x,y)= {x_value} and {y_value}")
print(f"value of langrange1 and langrange2= {l1} and {l2}")
print(f"minimized value={value}")

#plotting the functions
X=np.linspace(-6,6,400)
Y=np.linspace(-6,6,400)
X,Y=np.meshgrid(X,Y)  # creates 2D grid for contour
G1=X**2+4*Y**2-24
G2=X-Y

plt.figure(figsize=(6,6))
c1=plt.contour(X,Y,G1,colors="blue") # Contour for ellipse
c2=plt.contour(X,Y,G2,colors="green") # Contour for line

x_value,y_value,l1,l2,value=kkt()
plt.plot(x_value,y_value,color="red",marker="X",label='min value') #optimal point

plt.title("Constraints vs optimal point")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.savefig(r"C:\Users\Apoorva Singh\OneDrive - Amity University\Desktop\ANM Python assignment\Ass 2\Q3b.png", dpi=600)
plt.show()
```
